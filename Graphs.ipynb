```python
from collections import deque

graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "G"],
    "F": ["C"],
    "G": ["E"],
}

def bfs(graph, start):
  visited = set()

  queue = deque([start])

  while queue:
    node = queue.popleft()
    if node not in visited:
      print (node)
      visited.add(node)

      for neighbor in graph[node]:
        if neighbor not in visited:
          queue.append(neighbor)

bfs(graph, "A")
```

```python
#create a node
class Node:
  def __init__(self, value):
    self.value = value
    self.neighbors = []
    self.visited = False

#create the graph
class Graph:
  def __init__(self):
    self.nodes = []
  #adds nodes
  def add_node(self, node):
    self.nodes.append(node)
  #connect nodes
  def add_edge(self, node1, node2):
    node1.neighbors.append(node2)
    node2.neighbors.append(node1)

#build the graph
graph = Graph()
A = Node("A")
B = Node("B")
C = Node("C")
D = Node("D")
E = Node("E")
F = Node("F")
G = Node("G")

graph.add_node(A)
graph.add_node(B)
graph.add_node(C)
graph.add_node(D)
graph.add_node(E)
graph.add_node(F)
graph.add_node(G)

graph.add_edge(A,B)
graph.add_edge(A,C)
graph.add_edge(B,D)
graph.add_edge(B,E)
graph.add_edge(C,F)
graph.add_edge(E,G)

#make a queue
class Queue:
  def __init__(self):
    self.items = []
  def enqueue(self,item):
    self.items.append(item)
  def dequeue(self):
    return self.items.pop(0)
  def is_empty(self):
    return len(self.items) == 0

#BFS
def bfs(start):
  queue = Queue()
  start.visited = True
  queue.enqueue(start)
  while not queue.is_empty():
    current = queue.dequeue()
    print(current.value)
    for neighbor in current.neighbors:
      if not neighbor.visited:
        neighbor.visited = True
        queue.enqueue(neighbor)

#make a stack
class Stack:
  def __init__(self):
    self.items = []
  def push(self,item):
    self.items.append(item)
  def pop(self):
    return self.items.pop()
  def is_empty(self):
    return len(self.items)==0

#DFS
def dfs(start):
  stack = Stack()
  stack.push(start)
  while not stack.is_empty():
    current = stack.pop()
    if not current.visited:
      current.visited = True
      print(current.value)
      for neighbor in reversed(current.neighbors):
        if not neighbor.visited:
          stack.push(neighbor)

def reset_visited(graph):
  for node in graph.nodes:
    node.visited = False

bfs(A)
reset_visited(graph)
dfs(A)
```
